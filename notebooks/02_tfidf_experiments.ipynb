```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, f1_score
```

```python
data = pd.read_parquet('../data/processed/arxiv_text.parquet')
```

```python
vectorizer = TfidfVectorizer(max_features=10000, stop_words='english')
X = vectorizer.fit_transform([data['text'], str(data['year'])])
y = data['label']

print(vectorizer.get_feature_names_out())
print(X.shape)
```

### TF-IDF Baseline: Logistic Regression

```python
model = LogisticRegression(max_iter=1000)
model.fit(X, y)
```

### Evaluate Performance

```python
y_pred = model.predict(X)
accuracy = accuracy_score(y, y_pred)
f1 = f1_score(y, y_pred, average='macro')
print("Shape: ", X.shape)
print(f"Accuracy: {accuracy:.4f}")
print(f"F1-macro: {f1:.4f}")
```

### Test TF-IDF with n-grams

```python
print("ngram_range=(1,2)")
vectorizer.ngram_range = (1,2)
X = vectorizer.fit_transform(data['text'])
model.fit(X, y)
y_pred = model.predict(X)
accuracy = accuracy_score(y, y_pred)
f1 = f1_score(y, y_pred, average='macro')
print("Shape: ", X.shape)
print(f"Accuracy: {accuracy:.4f}")
print(f"F1-macro: {f1:.4f}")
```
